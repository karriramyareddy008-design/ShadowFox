```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("delhiaqi (2).csv")
df.head()
```

```python
df.columns
```

```python
df.info()
df.describe()
```

```python
df.columns
```

```python
df["date"] = pd.to_datetime(df["date"])
df.head()
```

```python
df["Month"] = df["date"].dt.month
df.head()
```

```python
monthly_avg = df.groupby("Month")["pm2_5"].mean()
monthly_avg
```

```python
import matplotlib.pyplot as plt

monthly_avg.plot(kind="bar")
plt.title("Monthly Average PM2.5 in Delhi")
plt.xlabel("Month")
plt.ylabel("Average PM2.5")
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,6))
sns.heatmap(df.corr(), annot=True)
plt.title("Correlation Between Pollutants")
plt.show()
```

```python
df.sort_values(by="pm2_5", ascending=False).head()

```

```python
plt.figure(figsize=(10,5))
plt.plot(df["date"], df["pm2_5"], label="PM2.5")
plt.plot(df["date"], df["pm10"], label="PM10")
plt.legend()
plt.title("PM2.5 vs PM10 Trend")
plt.xlabel("Date")
plt.ylabel("Pollution Level")
plt.show()
```

## Conclusion

The Delhi AQI analysis shows significant seasonal variation in air pollution levels.
PM2.5 and PM10 are major contributors to air pollution.
Higher pollution levels are observed in winter months, likely due to weather conditions and reduced air circulation.
Proper environmental control measures are required to reduce pollution levels.
